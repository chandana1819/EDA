```python
import numpy as np
np.random.seed(2)
arr = np.random.randint(10, size=(2,5))
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.nbytes)
```

```python
import numpy as np
np.random.seed(1)
arr = np.random.randint(10,size=(10,5))
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
print(arr.nbytes)
print(arr[-1,-2])
```

```python
import numpy as np
np.random.seed()
arr = np.random.randint(10, size=10)
print(arr)
print(arr[3])
print(arr[-2])
print(arr[3:9:1])
print(arr[::-1])
```
