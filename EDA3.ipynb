```python
import numpy as np

name = ["alice", "bob", "cathy", "jaan"]
age = [25, 45, 37, 19]
weight = [55.0, 85.5, 68.0, 29.5]

dtype = np.dtype({
    "names": ["name", "age", "weight"],
    "formats": ["U10", "i4", "f8"],
})

data = np.zeros(4, dtype=dtype)

print("Structured array using a compound data type")

data["name"] = name
data["age"] = age
data["weight"] = weight

print("Filled structured array")
print(data)

print("All names")
print(data["name"])

print("First row data")
print(data[0])

print("Name from the last row")
print(data[-1]["name"])

print("Names where age is under 30")
print(data[data["age"] < 30]["name"])
```
