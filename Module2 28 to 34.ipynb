```python
import pandas as pd
import numpy as np
```

```python
series= pd.Series([45,67,89, 23,56])
print(series)
print(series.values)
print(series.index)
```

```python
population = pd.Series([12_500_000, 11_000_000, 8_500_000, 7_000_000, 4_500_000],
    index=["Mumbai", "Delhi","Bengaluru","Chennai","Kolkata"])
print(population["Bengaluru"])
print(population["Delhi":"Chennai"])
```

```python
prices ={"Laptop":75000,"Phone":30000,"Tablet": 25000}
product_prices =pd.Series(prices)
print(product_prices)
print(product_prices.index)
print(product_prices.index.equals(pd.Index(prices.keys())))
```

```python
area = pd.Series([307713, 342239, 191791, 130058, 308252],
    index=["Rajasthan", "Madhya Pradesh","Maharashtra","Uttar Pradesh","Gujarat"])
population = pd.Series([81, 73, 112, 200, 60],
    index=area.index)
states = pd.DataFrame({"area": area,"population": population})
states["density"] = states["population"] / states["area"]
print(states)
```

```python
array = np.random.random((3,3))
df = pd.DataFrame(array,columns=["A", "B", "C"],index=["x", "y", "z"])
print(df)
```

```python
indA = pd.Index([2,3,5,7,11])
indB= pd.Index([1,3,5,7,9])
print("Intersection:", indA.intersection(indB))
print("Union:",indA.union(indB))
print("Symmetric difference:", indA.symmetric_difference(indB))
```
