```python
import pandas as pd
import numpy as np
```

```python
data=pd.Series([1,np.nan,"hello"])
print(data.isnull())
print(data[data.isnull()])
```

```python
df= pd.DataFrame(np.arange(20).reshape(5,4), 
columns=["A", "B", "C", "D"])
df.iloc[0,1] =np.nan
df.iloc[1,2] =np.nan
df.iloc[3,0] =np.nan
df.iloc[4,3] =np.nan
print(df)
print(df.isnull().sum())
```

```python
print(df.dropna())  
print(df.dropna(axis="columns"))
```

```python
df_empty = pd.DataFrame({ "A": [np.nan, 1, np.nan],"B": [np.nan, np.nan, 3],"C": [np.nan, 2, 4]})
print(df_empty.dropna(how="all"))
print(df.dropna(thresh=3))
```

```python
s = pd.Series([1,np.nan,2,None,3,np.nan])
print(s.fillna(0))
print(s.ffill())
print(s.bfill())
```

```python
temps = pd.DataFrame({"temperature":[20,22,np.nan,26,np.nan,30,32]})
print(temps.fillna(temps["temperature"].mean()))
print(temps.interpolate())
```
